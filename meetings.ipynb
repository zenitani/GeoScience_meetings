```python
from openpyxl import load_workbook
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.family'] = 'Serif'
plt.rcParams["font.size"] = 20
plt.rcParams['lines.linewidth'] = 2
plt.rcParams['lines.markersize'] = 7
plt.rcParams['figure.figsize'] = (6.4,4.8)
```

```python
wb = load_workbook(filename='meetings.xlsx', read_only=True)
ws = wb['Sheet1']
data = np.ndarray([30,9])
data[:,:] = None
for i, row in enumerate( ws.iter_rows(min_row=2) ):
    if( row[0].value == None ):
        break
    for j in range( 9 ):
        data[i,j] = row[j].value
```

```python
myalpha = 0.25
fig = plt.figure( figsize=(15,5) )
eguv= plt.bar( data[:,0]+0.1, data[:,3], width=0.2, color='blue', alpha=myalpha )
egu = plt.bar( data[:,0]+0.1, data[:,4], width=0.2, color='blue' )
jpguv= plt.bar( data[:,0]+0.3, data[:,5], width=0.2, color='magenta', alpha=myalpha )
jpgu = plt.bar( data[:,0]+0.3, data[:,6], width=0.2, color='magenta' )
aogsv= plt.bar( data[:,0]+0.5, data[:,7], width=0.2, color='green', alpha=myalpha )
aogs = plt.bar( data[:,0]+0.5, data[:,8], width=0.2, color='green' )
aguv= plt.bar( data[:,0]+0.7, data[:,1], width=0.2, color='red', alpha=myalpha )
agu = plt.bar( data[:,0]+0.7, data[:,2], width=0.2, color='red' )
plt.grid(linestyle='dotted', color='gray', lw=2, alpha=0.5)
plt.xlim( 2003, 2027)
plt.xlabel("Year")
plt.xticks( range(2005, 2030, 5) )
plt.ylim(0, 33000)
plt.ylabel("Participants")
plt.legend( (agu, egu, jpgu, aogs), ('AGU', 'EGU', 'JpGU', 'AOGS'), loc='upper left', fontsize=18, shadow=True)
plt.tight_layout()
# plt.savefig('meetings.png')
plt.show()
```
